```python
import pandas as pd
import plotly.express as px
import plotly.io as pio

# Carrega o CSV
df = pd.read_csv("/home/kayllany.oliveira/remote-repos/CellViability/resultados/cellpose/detalhado_CP_A.csv")

# Criar gráfico interativo com nome da imagem no hover
fig = px.scatter(
    df,
    x="area",
    y="eccentricity",
    title="Tamanho (Área) x Redondeza (eccentricidade) - cellpose_nuclei",
    labels={"area": "Área das células", "eccentricity": "Eccentricidade das células"},
    opacity=0.7,
    hover_data=["arquivo"],  # <- ajuste aqui o nome da coluna certa no seu CSV
)

# Mostrar no navegador
fig.show()

# Salvar como HTML interativo
pio.write_html(
    fig,
    file="/home/kayllany.oliveira/remote-repos/CellViability/resultados/cellpose/grafico_area_eccentricidade_CP_A.html",
    auto_open=False
)

```
