```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Caminhos
csv_path = "/home/kayllany.oliveira/remote-repos/CellViability/resultados/cellpose_sam/resumo_CP_4.csv"
txt_path = "/home/kayllany.oliveira/remote-repos/CellViability/resultados/columbus/2022-04-11T161022Z[10923]Contagem nucleo JSmetana Contras++[562178].result.1.txt"

# Lê CSV
df_csv = pd.read_csv(csv_path)
y = df_csv["total_objetos"].values

# Lê TXT separado por TAB
df_txt = pd.read_csv(txt_path, sep="\t")
x = df_txt["Nuclei Selected - Number of Objects"].values

# Ordena os dados para QQ-plot
x_quantiles = np.sort(x)
y_quantiles = np.sort(y)

# Ajusta tamanho para igualar (usando o menor tamanho entre os dois)
n = min(len(x_quantiles), len(y_quantiles))
x_quantiles = x_quantiles[:n]
y_quantiles = y_quantiles[:n]

# Cria figura
plt.figure(figsize=(6,6))

# Pontos personalizados
plt.scatter(x_quantiles, y_quantiles,
            s=15,           # tamanho das bolinhas
            facecolor= "skyblue" ,
            edgecolor='black',
            linewidth=0.5)

# Linha de referência 45°
lims = [min(x_quantiles.min(), y_quantiles.min()), max(x_quantiles.max(), y_quantiles.max())]
plt.plot(lims, lims, 'r--', linewidth=1)  # linha 45° pontilhada vermelha

# Labels e título
plt.ylabel("Cellpose_sam")
plt.xlabel("Columbus")
plt.title("QQ-Plot:Cellpose_sam vs Columbus")

# Salva e mostra
plt.savefig("qqplot_Cellpose_sam_vs_columbus.png", dpi=600, bbox_inches="tight")
plt.show()

```
